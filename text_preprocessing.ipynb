```python
from konlpy.tag import Komoran
from sklearn.feature_extraction.text import CountVectorizer

# Komoran 형태소 분석기 초기화
komoran = Komoran()

# Komoran 기반 tokenizer 함수 정의
def komoran_tokenizer(text):
    # 형태소 단위로 나누기 (어간, 접사, 조사 포함)
    return komoran.morphs(text)

texts = [
    "정부가 내년 최저임금을 3.5% 인상하기로 발표했다.",
    "최저임금이 내년에 3.5% 오를 예정이다.",
    "노동계는 최저임금 인상률이 너무 낮다고 반발하고 있다.",
    "손흥민이 오늘 경기에서 멀티골을 기록하며 팀의 승리를 이끌었다.",
    "토트넘은 손흥민의 활약으로 2대1 승리를 거두었다.",
    "오늘 경기에서 손흥민이 두 골을 넣으며 결정적인 역할을 했다.",
    "전국적으로 폭염 특보가 발령되었으며, 체감 온도는 35도를 넘었다.",
    "오늘은 올여름 들어 가장 더운 날로, 전국 대부분 지역에 폭염 경보가 내려졌다.",
    "기상청은 내일도 폭염이 계속될 것이라고 예보했다.",
    "무더위로 인한 온열질환자가 증가하고 있다.",
]

vectorizer = CountVectorizer(tokenizer=komoran_tokenizer)
X = vectorizer.fit_transform(texts)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python

from konlpy.tag import Komoran
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import pandas as pd

# 데이터
texts = [
    "정부가 내년 최저임금을 3.5% 인상하기로 발표했다.",
    "최저임금이 내년에 3.5% 오를 예정이다.",
    "노동계는 최저임금 인상률이 너무 낮다고 반발하고 있다.",
    "손흥민이 오늘 경기에서 멀티골을 기록하며 팀의 승리를 이끌었다.",
    "토트넘은 손흥민의 활약으로 2대1 승리를 거두었다.",
    "오늘 경기에서 손흥민이 두 골을 넣으며 결정적인 역할을 했다.",
    "전국적으로 폭염 특보가 발령되었으며, 체감 온도는 35도를 넘었다.",
    "오늘은 올여름 들어 가장 더운 날로, 전국 대부분 지역에 폭염 경보가 내려졌다.",
    "기상청은 내일도 폭염이 계속될 것이라고 예보했다.",
    "무더위로 인한 온열질환자가 증가하고 있다.",
]

# 불용어 목록
stop_words = ['이', '가', '을', '를', '에', '에서', '으로', '으로서', '이다', '하고', '하며', '도', '으로써',
              '은', '는', '께서', '보다', '까지', '부터', '이나', '라도', '든지', '조차', '마저']

# Komoran 객체 생성
komoran = Komoran()

# 사용자 정의 토크나이저
def tokenize(text):
    return [token for token in komoran.morphs(text) if token not in stop_words]

# CountVectorizer 벡터화
vectorizer = CountVectorizer(tokenizer=tokenize)
X = vectorizer.fit_transform(texts).toarray()

# 유사도 계산
similarity = cosine_similarity(X)

# t-SNE 차원 축소
tsne = TSNE(n_components=2, random_state=0, perplexity=5)
X_tsne = tsne.fit_transform(X)

# 시각화
plt.figure(figsize=(10, 8))
for i, label in enumerate(texts):
    x, y = X_tsne[i]
    plt.scatter(x, y)
    plt.annotate(f"{i+1}", (x, y), fontsize=12)
plt.title("문장 시각화 (t-SNE + Komoran)", fontsize=14)
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.grid(True)
plt.tight_layout()
plt.show()

# 유사도 행렬 출력
similarity_df = pd.DataFrame(similarity,
                             index=[f"문장{i+1}" for i in range(len(texts))],
                             columns=[f"문장{i+1}" for i in range(len(texts))])
print(similarity_df.round(2))
```
